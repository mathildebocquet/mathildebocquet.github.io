```python
import numpy as np
import random as rd #on importe les modules qui permettent de tirer aléatoirement

#Paramètres de l'expérience:
N=10000
M=1000
p=M/N
n=1000
X=[0 for i in range(N)] #on numérote implicitement les manchôts de 1 à N, X est l'ensemble des manchots

#Capture:
capture=rd.sample(range(1, N+1), M) #on tire au hasard M manchots, ce sont les manchots capturés

#Marquage:
for k in capture:
  X[k-1]=1 #On associe la valeur 1 aux manchots marqués

#Recapture
numrecapture=rd.sample(range(1,N+1),n) #on tire au hasard n manchots, ce sont les manchots recapturés
recapture=[X[k-1] for k in numrecapture]
m=np.sum(recapture) #on compte combien d'entre eux sont marqués
f=m/n

print("Une estimation du nombre de manchôts est",M/f)

#Calcul de l'intervalle de confiance sur N
borneinf=(n**0.5*M)/(f*n**0.5+1)
bornesup=(n**0.5*M)/(f*n**0.5-1)
print("Un intervalle de confiance à 95% sur N est [",borneinf,",",bornesup,"]")
```
